# Practical 01 – Smart Home Controller Intelligent Agent

```python
def smart_home(temperature, light_level, motion):
    fan = 'ON' if temperature > 28 else 'OFF'
    heater = 'ON' if temperature < 18 else 'OFF'
    light = 'ON' if motion and light_level < 40 else 'OFF'
    return {'Fan': fan, 'Heater': heater, 'Light': light}
for temperature, light, motion in [(32, 20, True), (15, 70, False), (24, 10, True)]:
    print('Sensors:', temperature, light, motion, 'Actions:', smart_home(temperature, light, motion))
```
