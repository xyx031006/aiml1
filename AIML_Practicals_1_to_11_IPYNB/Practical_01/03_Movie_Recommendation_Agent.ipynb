# Practical 01 – Movie Recommendation Intelligent Agent

```python
movies = {'action': ['Mad Max: Fury Road', 'Mission: Impossible'], 'comedy': ['The Intern', 'Paddington'], 'science fiction': ['Interstellar', 'The Martian']}

def recommend(genre):
    return movies.get(genre.lower(), ['No matching genre. Try: ' + ', '.join(movies)])
for genre in ['action', 'comedy', 'science fiction']:
    print(genre, ':', ', '.join(recommend(genre)))
```
