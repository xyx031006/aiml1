# Practical 01 – Rule-Based Chatbot Intelligent Agent

```python
def chatbot(message):
    text = message.lower().strip()
    if any((word in text for word in ['hello', 'hi', 'hey'])):
        return 'Hello! How can I help?'
    if 'name' in text:
        return 'I am a rule-based intelligent agent.'
    if 'time' in text:
        from datetime import datetime
        return 'Current time: ' + datetime.now().strftime('%H:%M')
    if 'help' in text:
        return 'Ask about my name or the time; type exit to stop.'
    return 'I do not understand. Type help for options.'
print('Chatbot Agent (type exit to stop)')
while True:
    message = input('You: ')
    if message.lower().strip() in ('exit', 'quit'):
        break
    print('Bot:', chatbot(message))
```
