# Practical 03 – A* Search

```python
import heapq
from time import perf_counter
graph = {'S': {'A': 2, 'B': 1}, 'A': {'G': 3}, 'B': {'C': 1, 'G': 8}, 'C': {'G': 3}, 'G': {}}
heuristic = {'S': 4, 'A': 3, 'B': 1, 'C': 1, 'G': 0}
```

```python
def astar(start, goal):
    frontier = [(heuristic[start], 0, start, [start])]
    best = {start: 0}
    explored = 0
    while frontier:
        _, cost, node, path = heapq.heappop(frontier)
        if cost > best[node]:
            continue
        explored += 1
        if node == goal:
            return (path, cost, explored)
        for neighbor, weight in graph[node].items():
            new_cost = cost + weight
            if new_cost < best.get(neighbor, float('inf')):
                best[neighbor] = new_cost
                heapq.heappush(frontier, (new_cost + heuristic[neighbor], new_cost, neighbor, path + [neighbor]))
    return (None, None, explored)
print('A* path, cost, explored:', astar('S', 'G'))
```
