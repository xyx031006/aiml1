# Practical 03 – Greedy Best-First Search and A* Combined

```python
import heapq
from time import perf_counter
graph = {'S': {'A': 2, 'B': 1}, 'A': {'G': 3}, 'B': {'C': 1, 'G': 8}, 'C': {'G': 3}, 'G': {}}
heuristic = {'S': 4, 'A': 3, 'B': 1, 'C': 1, 'G': 0}
```

```python
def search(method):
    frontier = [(heuristic['S'], 0, 'S', ['S'])]
    best = {'S': 0}
    visited = set()
    explored = 0
    while frontier:
        _, cost, node, path = heapq.heappop(frontier)
        if method == 'Greedy':
            if node in visited:
                continue
            visited.add(node)
        elif cost > best.get(node, float('inf')):
            continue
        explored += 1
        if node == 'G':
            return (path, cost, explored)
        for neighbor, weight in graph[node].items():
            nc = cost + weight
            if method == 'Greedy':
                if neighbor not in visited:
                    heapq.heappush(frontier, (heuristic[neighbor], nc, neighbor, path + [neighbor]))
            elif nc < best.get(neighbor, float('inf')):
                best[neighbor] = nc
                heapq.heappush(frontier, (nc + heuristic[neighbor], nc, neighbor, path + [neighbor]))
    return (None, None, explored)
for method in ['Greedy', 'A*']:
    t = perf_counter()
    print(method, search(method), 'Seconds:', perf_counter() - t)
```
