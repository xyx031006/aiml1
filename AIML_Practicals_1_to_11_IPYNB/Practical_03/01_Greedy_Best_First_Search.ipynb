# Practical 03 – Greedy Best-First Search

```python
import heapq
from time import perf_counter
graph = {'S': {'A': 2, 'B': 1}, 'A': {'G': 3}, 'B': {'C': 1, 'G': 8}, 'C': {'G': 3}, 'G': {}}
heuristic = {'S': 4, 'A': 3, 'B': 1, 'C': 1, 'G': 0}
```

```python
def greedy(start, goal):
    frontier = [(heuristic[start], start, [start], 0)]
    visited = set()
    explored = 0
    while frontier:
        _, node, path, cost = heapq.heappop(frontier)
        if node in visited:
            continue
        visited.add(node)
        explored += 1
        if node == goal:
            return (path, cost, explored)
        for neighbor, weight in graph[node].items():
            if neighbor not in visited:
                heapq.heappush(frontier, (heuristic[neighbor], neighbor, path + [neighbor], cost + weight))
    return (None, None, explored)
print('Greedy path, cost, explored:', greedy('S', 'G'))
```
