# Practical 02 – Depth First Search (DFS)

```python
graph = {'A': ['B', 'C'], 'B': ['D', 'E'], 'C': ['F'], 'D': ['G'], 'E': ['G'], 'F': ['G'], 'G': []}
from time import perf_counter
```

```python
def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = set()
    explored = 0
    while stack:
        node, path = stack.pop()
        if node in visited:
            continue
        visited.add(node)
        explored += 1
        if node == goal:
            return (path, explored)
        for neighbor in reversed(graph.get(node, [])):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))
    return (None, explored)
```

```python
t = perf_counter()
path, nodes = dfs(graph, 'A', 'G')
print('DFS Path:', path, 'Length:', len(path) - 1 if path else None, 'Nodes explored:', nodes, 'Seconds:', perf_counter() - t)
```
