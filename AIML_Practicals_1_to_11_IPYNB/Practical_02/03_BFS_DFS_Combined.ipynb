# Practical 02 – BFS and DFS Combined Comparison

```python
graph = {'A': ['B', 'C'], 'B': ['D', 'E'], 'C': ['F'], 'D': ['G'], 'E': ['G'], 'F': ['G'], 'G': []}
from collections import deque
from time import perf_counter

def search(method):
    frontier = deque([('A', ['A'])])
    visited = set()
    explored = 0
    while frontier:
        node, path = frontier.popleft() if method == 'BFS' else frontier.pop()
        if node in visited:
            continue
        visited.add(node)
        explored += 1
        if node == 'G':
            return (path, explored)
        neighbors = graph.get(node, []) if method == 'BFS' else list(reversed(graph.get(node, [])))
        for n in neighbors:
            if n not in visited:
                frontier.append((n, path + [n]))
    return (None, explored)
for method in ['BFS', 'DFS']:
    t = perf_counter()
    path, nodes = search(method)
    print(method, 'Path:', path, 'Length:', len(path) - 1 if path else None, 'Explored:', nodes, 'Seconds:', perf_counter() - t)
```
