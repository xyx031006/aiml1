# Practical 04 – Backward Chaining Expert System

```python
facts = {'attendance_ok', 'assignments_complete', 'fees_paid'}
rules = [({'attendance_ok', 'assignments_complete'}, 'academically_eligible'), ({'academically_eligible', 'fees_paid'}, 'hall_ticket_eligible')]

def backward(goal, facts, rules, seen=None):
    if goal in facts:
        return True
    seen = set() if seen is None else seen
    if goal in seen:
        return False
    seen.add(goal)
    return any((all((backward(condition, facts, rules, seen.copy()) for condition in conditions)) for conditions, conclusion in rules if conclusion == goal))
print('Goal: hall_ticket_eligible')
print('Proved:', backward('hall_ticket_eligible', facts, rules))
```
