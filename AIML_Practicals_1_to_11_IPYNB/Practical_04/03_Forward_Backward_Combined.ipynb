# Practical 04 – Forward and Backward Chaining Combined

```python
facts = {'attendance_ok', 'assignments_complete', 'fees_paid'}
rules = [({'attendance_ok', 'assignments_complete'}, 'academically_eligible'), ({'academically_eligible', 'fees_paid'}, 'hall_ticket_eligible')]
```

```python
def forward(facts, rules):
    known = set(facts)
    while True:
        new = {head for body, head in rules if body <= known} - known
        if not new:
            return known
        known.update(new)
```

```python
def backward(goal, facts, rules, seen=None):
    if goal in facts:
        return True
    seen = set() if seen is None else seen
    if goal in seen:
        return False
    seen.add(goal)
    return any((all((backward(p, facts, rules, seen.copy()) for p in body)) for body, head in rules if head == goal))
```

```python
print('Forward:', 'hall_ticket_eligible' in forward(facts, rules))
print('Backward:', backward('hall_ticket_eligible', facts, rules))
```
