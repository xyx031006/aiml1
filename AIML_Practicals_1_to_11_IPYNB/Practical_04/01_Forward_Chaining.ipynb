# Practical 04 – Forward Chaining Expert System

```python
facts = {'attendance_ok', 'assignments_complete', 'fees_paid'}
rules = [({'attendance_ok', 'assignments_complete'}, 'academically_eligible'), ({'academically_eligible', 'fees_paid'}, 'hall_ticket_eligible')]

def forward(facts, rules):
    known = set(facts)
    changed = True
    while changed:
        changed = False
        for conditions, conclusion in rules:
            if conditions <= known and conclusion not in known:
                known.add(conclusion)
                changed = True
                print('Derived:', conclusion)
    return known
print('Hall ticket eligible:', 'hall_ticket_eligible' in forward(facts, rules))
```
